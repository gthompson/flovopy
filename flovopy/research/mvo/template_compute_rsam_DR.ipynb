# RSAM and reduced displacement

You can now request just RSAM and DR:

````python
from flovopy.processing.archive_metrics import Config, run

config = Config(
    sds_root="/data/Montserrat/SDS",
    output_root="/data/Montserrat/SAM",
    start="2000-01-01",
    end="2000-01-08",

    products=("RSAM", "DR"),

    stationxml="/data/Montserrat/stations.xml",

    source_latitude=16.72,
    source_longitude=-62.18,
    source_elevation=1000.0,

    sampling_interval=60,
    workers=4,
)

summary = run(config)
````

Command line version:

```
python -m flovopy.processing.archive_metrics \
    --sds-root /data/Montserrat/SDS \
    --output-root /data/Montserrat/SAM \
    --start 2000-01-01 \
    --end 2000-01-08 \
    --products RSAM DR \
    --stationxml /data/Montserrat/stations.xml \
    --source-latitude 16.72 \
    --source-longitude -62.18 \
    --source-elevation 1000 \
    --workers 4
```
